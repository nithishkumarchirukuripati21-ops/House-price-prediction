```python
import shap
import joblib
import pandas as pd
import matplotlib.pyplot as plt
```

```python
model = joblib.load(
    "../models/best_xgboost_model.pkl"
)
data = pd.read_csv(
    "../data/preprocessed_housing.csv"
)
feature_columns = joblib.load(
    "../models/feature_columns.pkl"
)
```

```python
X = data.drop(
    "SalePrice",
    axis=1
)


X = X[feature_columns]


X.shape
```

```python
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X)
type(shap_values)
shap.summary_plot(
    shap_values,
    X
)
```

```python
import joblib

joblib.dump(
    explainer,
    "../models/shap_explainer.pkl"
)
```
