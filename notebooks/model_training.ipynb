```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import(
    mean_absolute_error,
    mean_squared_error,
    r2_score
) 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("../data/preprocessed_housing.csv")
df.head()
```

```python
x=df.drop(columns=["Id","SalePrice"])
y=df["SalePrice"]
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
scalar=StandardScaler()
X_trained_scaled=scalar.fit_transform(X_train)
X_test_scaled=scalar.transform(X_test)
```

```python
results = []
```

LINEAR REGRESSION MODEL

```python
lr_model=LinearRegression()
lr_model.fit(X_trained_scaled,Y_train)
y_pred_lr=lr_model.predict(X_test_scaled)
```

```python
mae_lr=mean_absolute_error(Y_test,y_pred_lr)
mse_lr=mean_squared_error(Y_test,y_pred_lr)
rmse_lr=mse**0.5
r2_lr=r2_score(Y_test,y_pred_lr)
print(f"MAE : {mae_lr:.2f}")
print(f"MSE : {mse_lr:.2f}")
print(f"RMSE : {rmse_lr:.2f}")
print(f"R^2 : {r2_lr:.4f}")
```

```python
results.append({
    "Model": "Linear Regression",
    "MAE": mae_lr,
    "RMSE": rmse_lr,
    "R2": r2_lr
})
```

DECISION TREE MODEL

```python
dt_model = DecisionTreeRegressor(random_state=42)
dt_model.fit(X_train,Y_train)
y_pred_dt=dt_model.predict(X_test)
```

```python
mae_dt = mean_absolute_error(Y_test, y_pred_dt)
mse_dt = mean_squared_error(Y_test, y_pred_dt)
rmse_dt = mse ** 0.5
r2_dt = r2_score(Y_test, y_pred_dt)
print(f"MAE : {mae_dt:.2f}")
print(f"MSE : {mse_dt:.2f}")
print(f"RMSE : {rmse_dt:.2f}")
print(f"R²   : {r2_dt:.4f}")
```

```python
results.append({
    "Model": "Decision Tree",
    "MAE": mae_dt,
    "RMSE": rmse_dt,
    "R2": r2_dt
})
```

RANDOM FOREST

```python
rf_model = RandomForestRegressor(random_state=42)
rf_model.fit(X_train,Y_train)
y_pred_rf=rf_model.predict(X_test)
```

```python
mae_rf = mean_absolute_error(Y_test, y_pred_rf)
mse_rf = mean_squared_error(Y_test, y_pred_rf)
rmse_rf = mse_rf ** 0.5
r2_rf = r2_score(Y_test, y_pred_rf)

print(f"MAE : {mae_rf:.2f}")
print(f"MSE : {mse_rf:.2f}")
print(f"RMSE: {rmse_rf:.2f}")
print(f"R²  : {r2_rf:.4f}")
```

```python
results.append({
    "Model": "Random Forest",
    "MAE": mae_rf,
    "RMSE": rmse_rf,
    "R2": r2_rf
})
```

XGBOOST

```python
xgb_model = XGBRegressor(
    random_state=42,
    objective="reg:squarederror"
)
xgb_model.fit(X_train, Y_train)
y_pred_xgb = xgb_model.predict(X_test)
```

```python
mae_xgb = mean_absolute_error(Y_test, y_pred_xgb)
mse_xgb = mean_squared_error(Y_test, y_pred_xgb)
rmse_xgb = mse_xgb ** 0.5
r2_xgb = r2_score(Y_test, y_pred_xgb)

print(f"MAE : {mae_xgb:.2f}")
print(f"MSE : {mse_xgb:.2f}")
print(f"RMSE: {rmse_xgb:.2f}")
print(f"R²  : {r2_xgb:.4f}")
```

```python
results.append({
    "Model": "XGBoost",
    "MAE": mae_xgb,
    "RMSE": rmse_xgb,
    "R2": r2_xgb
})
```

```python
results_df = pd.DataFrame(results)
results_df.sort_values(by="R2", ascending=False).reset_index(drop=True)
```
